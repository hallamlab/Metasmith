```python
from metasmith.python_api import *
from local.constants import WORKSPACE_ROOT
```

```python
mover = Logistics()
dest = Source.FromLocal(WORKSPACE_ROOT/"data/uniprot_sprot.fasta.gz")

if not dest.GetPath().exists():
    mover.QueueTransfer(
        src = HttpSource.Parse("https://ftp.uniprot.org/pub/databases/uniprot/current_release/knowledgebase/complete/uniprot_sprot.fasta.gz").AsSource(),
        dest = dest,
    )
    mover.ExecuteTransfers()

# remember to unzip the file...
swiss_prot = dest.GetPath().with_suffix("")
swiss_prot.relative_to(WORKSPACE_ROOT)
```

```python
from Bio import SeqIO

to_keep = []
for e in SeqIO.parse(swiss_prot, "fasta"):
    if e.id not in WL: continue
    to_keep.append(e)

with open(WORKSPACE_ROOT/"src/metasmith/example_resources/swissprot_bcaa.fna", "w") as f:
    SeqIO.write(to_keep, f, "fasta")
```
