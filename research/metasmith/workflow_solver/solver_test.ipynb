```python
def _print_sol(sol: Solution):
    print(f"complete: {sol.complete}, variants: {len(sol.dependency_plans)}")
    for i, plan in enumerate(sol.dependency_plans):
        print(f">>> variant {i+1}")
        for j, step in enumerate(plan):
            print(f"step {j}")
            print(step.transform)
            print("used")
            for x in step.used:
                print(" ", x)
            print("produced")
            for x in step.produced:
                print(" ", x)
            # if len(step.deleted) > 0:
            #     print("deleted")
            #     for x in step.deleted:
            #         print(" ", x)
            print()

def _set(s: str):
    return set(s.split(", "))
```
